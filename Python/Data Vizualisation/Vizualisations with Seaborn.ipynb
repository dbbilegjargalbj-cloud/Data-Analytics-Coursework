```python
housing_filtered = housing[housing['state'].isin(['Massachusetts', 'Puerto Rico', 'New York'])]
```

```python
sns.barplot(housing_filtered, x = 'state', y = 'price', estimator = 'median')
plt.plot()
```

## Strip plot, swarm plot and violin plot

```python
sns.stripplot(x = 'species', y = 'sepal_length', data = iris, hue = 'species')
```

```python
sns.swarmplot(x = 'species', y = 'sepal_length', data = iris, hue = 'species')
```

```python
sns.violinplot(x = 'species', y = 'sepal_length', data = iris, hue = 'species')
```
