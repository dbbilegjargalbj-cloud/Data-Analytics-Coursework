# Numpy Arrays and Pandas Series

```python
import numpy as np
import pandas as pd
```

```python
my_list = [10,25,35]
```

```python
# Creating Numpy Arrays (index is hidden)
```

```python
x = np.array(my_list)
print(x)
```

```python
type(x)
```

```python
# Creating Pandas Series (index is shown)
```

```python
pd.Series(my_list)
```

```python
y = pd.Series(my_list)
```

```python
type(y)
```

```python
# In Pandas Series you can name indexes
```

```python
index_values= ['Best', 'Second Best', 'Third Best']
my_list = [10,25,35]

one = pd.Series(data = my_list, index = index_values)
```

```python
index_values2 = ['Best', 'Second Best', 'Forth Best']
my_list2 = [10,25,40]

two = pd.Series(data = my_list2, index = index_values2)
```

```python
one
```

```python
two
```

```python
total = one + two
total
```

```python
# can refer to index using both the index name and number
```

```python
total['Best']
```

```python
total[0]
```
