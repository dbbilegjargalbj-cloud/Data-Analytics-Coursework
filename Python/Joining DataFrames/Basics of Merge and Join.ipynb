```python
import pandas as pd
```

```python
data1 = {
    'employee_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Dwight Schrute', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson'],
    'job_title': ['Regional Manager', 'Sales Representative', 'Receptionist', 'Assistant to the Regional Manager', 'Head of Accounting', 'Accountant', 'Accountant', 'Sales Representative'],
    'salary': [100000, 60000, 55000, 65000, 60000, 55000, 55000, 60000]
}

df1 = pd.DataFrame(data1)

data2 = {
    'emp_id': [1, 2, 3, 5, 6, 7, 8, 9],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson', 'Phyllis Vance'],
    'department': ['Management', 'Sales', 'Reception', 'Accounting', 'Accounting', 'Accounting', 'Sales', 'Sales']
}

df2 = pd.DataFrame(data2)
```

```python
df1
```

```python
df2
```

```python
df1.merge(df2) # default performing automatically inner join
# but better to specify
```

```python
df1.merge(df2, how = 'left') #on df1
```

```python
df1.merge(df2, how = 'right') #on df2
```

```python
df1.merge(df2, how = 'outer') #takes everything from both data frames and every thats similar
```

```python
df1.merge(df2, how = 'outer', on = 'employee_id') # specify on what to join
```

```python
df1.merge(df2, how = 'outer', on = 'employee_id') # specify on what to join
```

```python
# lets change the df2.employee_id to emp_id..
# then we have to specify on which column for each dataframe using left_on and right_on
```

```python
df1.merge(df2, how = 'outer', left_on = 'employee_id', right_on = 'emp_id') # specify on what to join
```

```python
df1.join(df2, how = 'inner', lsuffix = '_df1', rsuffix = '_df2')
# gives us messy data  because it joins on indexes of df1 and df2
```

```python
# set_index to correct it

df1.set_index('employee_id', inplace = True)
df2.set_index('emp_id', inplace = True)
```

```python
df1.join(df2, how = 'inner', lsuffixs = '_df1', rsuffix = '_df2')
# gives us messy data  because it joins on indexes of df1 and df2
```
